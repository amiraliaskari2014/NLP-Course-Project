````python
ENSEMBLE_MODEL_IDS = {
    'qwen': 'Qwen/Qwen2.5-Math-7B',
    'llama': 'meta-llama/Llama-3.2-3B-Instruct',
}

# Clear leftovers from earlier failed loads before creating the model.
for _name in ['model', 'tokenizer', 'models', 'tokenizers']:
    if _name in globals():
        try:
            del globals()[_name]
        except Exception:
            pass
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

# Do not paste tokens in notebook output. Preferred: Colab Secrets -> HF_TOKEN.
def get_hf_token():
    token = os.environ.get('HF_TOKEN')
    if token:
        return token
    try:
        from google.colab import userdata
        token = userdata.get('HF_TOKEN')
        if token:
            return token
    except Exception:
        pass
    return getpass.getpass('Enter HF token for gated models (input hidden): ')

HF_TOKEN = get_hf_token()

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type='nf4',
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
)

tokenizers = {}
models = {}
ACTIVE_ROLE = None
PRELOAD_DONE = False
KEEP_MODELS_LOADED = False

def clear_model_from_memory(role):
    if role in models:
        del models[role]
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

def hard_clear_vram():
    for role in list(models.keys()):
        clear_model_from_memory(role)
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.ipc_collect()

def load_4bit_model(model_id):
    model_path = snapshot_download(repo_id=model_id, token=HF_TOKEN)
    tok = tokenizers.get(model_id)
    if tok is None:
        tok = AutoTokenizer.from_pretrained(model_path, token=HF_TOKEN)
        if tok.pad_token is None:
            tok.pad_token = tok.eos_token
        tokenizers[model_id] = tok
    mdl = AutoModelForCausalLM.from_pretrained(
        model_path,
        quantization_config=bnb_config,
        device_map={'': 0} if torch.cuda.is_available() else {'': 'cpu'},
        token=HF_TOKEN,
    )
    mdl.eval()
    return tok, mdl

def get_role_model(role='qwen'):
    global ACTIVE_ROLE
    if role in ['router', 'math', 'verifier']:
        role = 'qwen'
    if role not in ENSEMBLE_MODEL_IDS:
        role = 'qwen'

    if role in models:
        return tokenizers[role], models[role]

    if not KEEP_MODELS_LOADED and ACTIVE_ROLE is not None and ACTIVE_ROLE != role:
        clear_model_from_memory(ACTIVE_ROLE)

    model_id = ENSEMBLE_MODEL_IDS[role]
    tok, mdl = load_4bit_model(model_id)
    tokenizers[role], models[role] = tok, mdl
    ACTIVE_ROLE = role
    print('Loaded 4-bit model:', role, model_id)
    return tok, mdl

def print_vram(label='VRAM'):
    if not torch.cuda.is_available():
        print(label + ': CUDA is not available')
        return
    free, total = torch.cuda.mem_get_info()
    allocated = torch.cuda.memory_allocated()
    reserved = torch.cuda.memory_reserved()
    print(f'{label}: free={free/1024**3:.2f}GB total={total/1024**3:.2f}GB allocated={allocated/1024**3:.2f}GB reserved={reserved/1024**3:.2f}GB')

def preload_ensemble(warm_role='qwen', force=False, keep_loaded=True):
    global PRELOAD_DONE, KEEP_MODELS_LOADED, tokenizer, model
    if globals().get('PRELOAD_DONE') and not force:
        print('Ensemble preload already complete. Loaded roles:', sorted(models.keys()))
        return

    print('Preloading ensemble model files and tokenizers...')
    for role, model_id in ENSEMBLE_MODEL_IDS.items():
        model_path = snapshot_download(repo_id=model_id, token=HF_TOKEN)
        tok = tokenizers.get(model_id)
        if tok is None:
            tok = AutoTokenizer.from_pretrained(model_path, token=HF_TOKEN)
            if tok.pad_token is None:
                tok.pad_token = tok.eos_token
            tokenizers[model_id] = tok
        print('Cached tokenizer/files:', role, model_id)

    if keep_loaded:
        try:
            KEEP_MODELS_LOADED = True
            for role in ENSEMBLE_MODEL_IDS:
                get_role_model(role)
            PRELOAD_DONE = True
            print('Both 4-bit models are resident in VRAM:', sorted(models.keys()))
            print_vram('After dual preload')
            return
        except Exception as exc:
            print('Could not keep both models loaded; falling back to one active model.')
            print(type(exc).__name__ + ':', str(exc).split('\n')[0])
            KEEP_MODELS_LOADED = False
            for role in list(models.keys()):
                clear_model_from_memory(role)
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    if warm_role:
        tokenizer, model = get_role_model(warm_role)
        print('Warmed active model before game:', warm_role)

    PRELOAD_DONE = True
    print('Ensemble preload complete. Loaded roles:', sorted(models.keys()))
    print_vram('After preload')

# Models are loaded lazily one at a time by get_role_model(role).
tokenizer = None
model = None
print('Configured 4-bit ensemble models:', ENSEMBLE_MODEL_IDS)
````

````python
LETTERS = ['A', 'B', 'C', 'D']

SAFE_BUILTINS = {
    'abs': abs,
    'all': all,
    'any': any,
    'bool': bool,
    'complex': complex,
    'dict': dict,
    'enumerate': enumerate,
    'float': float,
    'int': int,
    'len': len,
    'list': list,
    'max': max,
    'min': min,
    'pow': pow,
    'print': print,
    'range': range,
    'round': round,
    'set': set,
    'sorted': sorted,
    'str': str,
    'sum': sum,
    'tuple': tuple,
    'zip': zip,
}

ALLOWED_IMPORT_NAMES = {
    'math',
    'fractions',
    'decimal',
    'itertools',
    'statistics',
    'functools',
    'sympy',
    'numpy',
}

def safe_import(name, globals=None, locals=None, fromlist=(), level=0):
    root_name = name.split('.')[0]
    if root_name in ALLOWED_IMPORT_NAMES:
        return importlib.import_module(name)
    raise ImportError(f'Import not allowed: {name}')

SAFE_BUILTINS['__import__'] = safe_import

MODE_MARGIN_FOR_PYTHON_VERIFY = 1.0

def format_question(question):
    lines = [f'Question: {question.text}']
    for i, option in enumerate(question.options[:4]):
        lines.append(f'{LETTERS[i]}) {option.text}')
    return '\n'.join(lines)

def preview(text, limit=500):
    text = str(text).replace('\n', ' | ').strip()
    if len(text) <= limit:
        return text
    return text[:limit - 3] + '...'

def generate_text(prompt, max_new_tokens=128, max_time=12, stop=None, role='router'):
    tok, mdl = get_role_model(role)
    inputs = tok(prompt, return_tensors='pt').to(mdl.device)
    input_len = inputs['input_ids'].shape[1]
    generation_kwargs = {
        'max_new_tokens': max_new_tokens,
        'do_sample': False,
        'pad_token_id': tok.eos_token_id,
    }
    if max_time is not None:
        generation_kwargs['max_time'] = max_time

    try:
        with torch.no_grad():
            outputs = mdl.generate(**inputs, **generation_kwargs)
    except torch.cuda.OutOfMemoryError:
        torch.cuda.empty_cache()
        return ''

    text = tok.decode(outputs[0][input_len:], skip_special_tokens=True).strip()
    if stop:
        for marker in stop:
            if marker in text:
                text = text.split(marker)[0].strip()
    return text

def extract_letter(text):
    text = str(text).strip().upper()
    text = re.sub(r'<\s*A\s*/\s*B\s*/\s*C\s*/\s*D\s*>', ' ', text)
    text = re.sub(r'A\s*/\s*B\s*/\s*C\s*/\s*D', ' ', text)
    patterns = [
        r'FINAL\s+ANSWER\s*[:\-]?\s*([ABCD])\b(?!\s*/)',
        r'ANSWER\s*[:\-]?\s*([ABCD])\b(?!\s*/)',
        r'OPTION\s*([ABCD])\b(?!\s*/)',
    ]
    for pattern in patterns:
        match = re.search(pattern, text)
        if match:
            return match.group(1)
    leading = re.match(r'^\s*([ABCD])\b', text)
    if leading:
        return leading.group(1)
    matches = re.findall(r'\b([ABCD])\b', text)
    if matches:
        return matches[-1]
    if text and text[0] in LETTERS:
        return text[0]
    return None

def option_to_number(text):
    cleaned = str(text).replace(',', '').replace('\u2212', '-')
    frac = re.search(r'[-+]?\d+\s*/\s*[-+]?\d+', cleaned)
    if frac:
        num, den = frac.group(0).split('/')
        try:
            return float(fractions.Fraction(int(num), int(den)))
        except Exception:
            pass
    nums = re.findall(r'[-+]?(?:\d+\.\d+|\d+|\.\d+)(?:e[-+]?\d+)?', cleaned, flags=re.IGNORECASE)
    if not nums:
        return None
    try:
        return float(nums[-1])
    except ValueError:
        return None

def letter_from_code_output(question, code_output):
    text = str(code_output).strip()
    if text.startswith('Python error:'):
        return None

    explicit = re.search(r'\b([ABCD])\b', text.upper())
    if explicit:
        return explicit.group(1)

    output_numbers = re.findall(r'[-+]?(?:\d+\.\d+|\d+|\.\d+)(?:e[-+]?\d+)?', text.replace(',', ''), flags=re.IGNORECASE)
    if not output_numbers:
        return None

    value = float(output_numbers[-1])
    option_numbers = [option_to_number(option.text) for option in question.options[:4]]
    numeric_options = [(i, n) for i, n in enumerate(option_numbers) if n is not None]
    if not numeric_options:
        return None

    best_i, best_n = min(numeric_options, key=lambda item: abs(item[1] - value))
    tolerance = max(1e-6, abs(value) * 1e-3)
    if abs(best_n - value) <= tolerance:
        return LETTERS[best_i]
    return None

def extract_python_code(text):
    text = str(text).strip()
    fenced = re.search(r'```(?:python)?\s*(.*?)```', text, flags=re.DOTALL | re.IGNORECASE)
    if fenced:
        return fenced.group(1).strip()

    lines = []
    for line in text.splitlines():
        stripped = line.strip()
        if stripped.startswith(('import ', 'from ', 'print(', 'answer', 'result', 'value', 'x', 'd', 'A', 'B')) or '=' in line or stripped.startswith(('for ', 'if ', 'def ', 'while ')):
            lines.append(line)
    return '\n'.join(lines).strip() if lines else text

def is_useful_python_code(code):
    code = extract_python_code(code)
    if not code.strip():
        return False
    active_lines = [line.strip() for line in code.splitlines() if line.strip() and not line.strip().startswith('#')]
    if not active_lines:
        return False
    joined = '\n'.join(active_lines)
    return any(token in joined for token in ['print(', 'for ', 'while ', 'if ', '=', 'math.', 'sympy', 'range(', 'gcd', 'det', 'Matrix'])

def run_python_code(code):
    code = extract_python_code(code)
    stdout = io.StringIO()
    namespace = {
        '__builtins__': SAFE_BUILTINS,
        'math': math,
        'fractions': fractions,
        'decimal': decimal,
        'itertools': itertools,
        'statistics': statistics,
        'functools': functools,
    }

    try:
        with contextlib.redirect_stdout(stdout):
            try:
                value = eval(code, namespace, namespace)
                if value is not None:
                    print(value)
            except SyntaxError:
                exec(code, namespace, namespace)
    except Exception as exc:
        return f'Python error: {type(exc).__name__}: {exc}'

    output = stdout.getvalue().strip()
    if output:
        return output

    for name in ['answer', 'result', 'value']:
        if name in namespace:
            return f'{name} = {namespace[name]}'
    return 'Python ran successfully but produced no output.'

def generate_python_solution(question):
    qtext = format_question(question)
    prompt = f'''Write executable Python 3 code to compute the mathematical facts needed for this A-D question.
Use only the shown A-D options. Do not invent E/F/G options.
Do not write comments, prose, or pseudo-code.
Do not directly guess an answer from words; compute values/truths using Python.
The code should print exactly one of these when possible:
- A, B, C, or D
- otherwise the computed numeric/exact value
Use standard library; sympy is allowed if useful.
Return only executable Python code, no markdown, no explanation.

{qtext}

Python code:'''
    raw = generate_text(prompt, max_new_tokens=160, max_time=14, stop=['\n\nQuestion:', '\n\nExplanation:'], role='math')
    code = extract_python_code(raw)
    if is_useful_python_code(code):
        return code, raw, 'primary_code_prompt'

    retry_prompt = f'''Write executable Python only. No comments. No explanation.
Compute the mathematical quantities/truth values needed, then print A/B/C/D or the computed value.
{qtext}

import math
from math import gcd
from fractions import Fraction
'''
    retry_raw = generate_text(retry_prompt, max_new_tokens=120, max_time=10, stop=['\n\n'], role='math')
    retry_code = extract_python_code(retry_raw)
    combined_raw = f'PRIMARY:\n{raw}\n\nRETRY:\n{retry_raw}'
    if is_useful_python_code(retry_code):
        return retry_code, combined_raw, 'retry_code_prompt'
    return '', combined_raw, 'no_useful_code_generated'

def parse_solution_plan(text):
    text = str(text).strip()
    mode_match = re.search(r'MODE\s*:\s*(DIRECT|PYTHON)', text, flags=re.IGNORECASE)
    answer_match = re.search(r'ANSWER\s*:\s*([ABCD]|UNKNOWN)\b', text, flags=re.IGNORECASE)
    code_match = re.search(r'CODE\s*:\s*(.*)', text, flags=re.DOTALL | re.IGNORECASE)
    mode = mode_match.group(1).upper() if mode_match else None
    answer = answer_match.group(1).upper() if answer_match else None
    code = extract_python_code(code_match.group(1).strip()) if code_match else ''
    if mode is None:
        mode = 'PYTHON' if code else 'DIRECT'
    if answer == 'UNKNOWN':
        answer = None
    return mode, answer, code

def constrained_mode_choice(question_text):
    prompt = f'''Select the best solving mode for this multiple-choice question.
PYTHON means code/calculation/symbolic computation would materially help.
DIRECT means conceptual knowledge or verbal reasoning is enough.
Answer with exactly one word: PYTHON or DIRECT.

{question_text}

Mode:'''
    try:
        tok, mdl = get_role_model('router')
        inputs = tok(prompt, return_tensors='pt').to(mdl.device)
        with torch.no_grad():
            logits = mdl(**inputs).logits[0, -1]
        scores = {}
        for mode in ['PYTHON', 'DIRECT']:
            candidate_scores = []
            for form in (mode, ' ' + mode):
                token_ids = tok.encode(form, add_special_tokens=False)
                if token_ids:
                    candidate_scores.append(float(logits[token_ids[0]].detach().cpu()))
            scores[mode] = max(candidate_scores) if candidate_scores else float('-inf')
        return max(scores, key=scores.get), scores
    except torch.cuda.OutOfMemoryError:
        torch.cuda.empty_cache()
        return 'DIRECT', {'error': 'CUDA OOM during mode choice'}
    except Exception as exc:
        return 'DIRECT', {'error': f'{type(exc).__name__}: {exc}'}

def generate_solution_plan(question):
    qtext = format_question(question)
    prompt = f'''Decide how to answer this A-D multiple-choice question.
Use DIRECT if the question is conceptual or can be answered without calculation.
Use PYTHON if code/calculation is useful.
Use only the shown A-D options. Do not invent E/F/G options.

Return exactly:
MODE: DIRECT or PYTHON
ANSWER: A/B/C/D or UNKNOWN
CODE: executable Python only if MODE is PYTHON, otherwise empty

{qtext}
'''
    raw = generate_text(prompt, max_new_tokens=160, max_time=12, stop=['\n\nQuestion:'], role='router')
    mode, answer, code = parse_solution_plan(raw)
    return mode, answer, code, raw

def direct_answer_fallback(question, code, code_output):
    qtext = format_question(question)
    prompt = f'''Choose the correct option using only A, B, C, or D.
Do not invent any other options.
Return exactly one of these four strings: Final Answer: A, Final Answer: B, Final Answer: C, Final Answer: D.

{qtext}

Python code attempted:
{code}

Python output:
{code_output}

Final Answer:'''
    raw = generate_text(prompt, max_new_tokens=24, max_time=6, role='verifier')
    return extract_letter(raw), raw

def constrained_letter_choice(question_text, context=''):
    prompt = f'''Choose the correct option. Valid answers are only A, B, C, or D.
{question_text}

{context}

Final Answer:'''
    try:
        tok, mdl = get_role_model('verifier')
        inputs = tok(prompt, return_tensors='pt').to(mdl.device)
        with torch.no_grad():
            logits = mdl(**inputs).logits[0, -1]
        scores = {}
        for letter in LETTERS:
            candidate_scores = []
            for form in (letter, ' ' + letter):
                token_ids = tok.encode(form, add_special_tokens=False)
                if token_ids:
                    candidate_scores.append(float(logits[token_ids[0]].detach().cpu()))
            scores[letter] = max(candidate_scores) if candidate_scores else float('-inf')
        return max(scores, key=scores.get), scores
    except torch.cuda.OutOfMemoryError:
        torch.cuda.empty_cache()
        return None, {'error': 'CUDA OOM during constrained choice'}
    except Exception as exc:
        return None, {'error': f'{type(exc).__name__}: {exc}'}

def format_repair_scores(scores):
    if not scores:
        return ''
    if isinstance(scores, dict) and 'error' in scores:
        return f"error={scores['error']}"
    return ', '.join(f'{letter}:{scores.get(letter, float("nan")):.3f}' for letter in LETTERS)

def format_mode_scores(scores):
    if not scores:
        return ''
    if isinstance(scores, dict) and 'error' in scores:
        return f"error={scores['error']}"
    return ', '.join(f'{mode}:{scores.get(mode, float("nan")):.3f}' for mode in ['PYTHON', 'DIRECT'])

def should_verify_direct_with_python(mode, mode_scores):
    if mode != 'DIRECT' or not isinstance(mode_scores, dict):
        return False
    if 'error' in mode_scores:
        return False
    python_score = mode_scores.get('PYTHON')
    direct_score = mode_scores.get('DIRECT')
    if python_score is None or direct_score is None:
        return False
    return direct_score - python_score <= MODE_MARGIN_FOR_PYTHON_VERIFY

def print_debug_details(details):
    print('Plan mode:', details.get('plan_mode'), '| Plan answer:', details.get('plan_answer'))
    if details.get('mode_scores'):
        print('Mode scores:', format_mode_scores(details['mode_scores']))
    if details.get('mode_override_reason'):
        print('Mode override:', details.get('mode_override_reason'))
    print('Code attempt:', details.get('code_attempt'))
    print('Generated code:', preview(details.get('generated_code', ''), 1000))
    print('Code raw:', preview(details.get('code_raw', ''), 1000))
    if details.get('code_output'):
        print('Code output:', preview(details['code_output'], 1000))
    if details.get('fallback_raw'):
        print('Fallback raw:', preview(details['fallback_raw'], 800))
    if details.get('repair_scores'):
        print('Repair scores:', format_repair_scores(details['repair_scores']))
    print('Selected letter:', details.get('selected_letter'), '| Reason:', details.get('selection_reason'))

def constrained_letter_choice_for_role(question_text, context='', role='qwen'):
    prompt = f'''Choose the correct option. Valid answers are only A, B, C, or D.
{question_text}

{context}

Final Answer:'''
    try:
        tok, mdl = get_role_model(role)
        inputs = tok(prompt, return_tensors='pt').to(mdl.device)
        with torch.no_grad():
            logits = mdl(**inputs).logits[0, -1]
        scores = {}
        for letter in LETTERS:
            candidate_scores = []
            for form in (letter, ' ' + letter):
                token_ids = tok.encode(form, add_special_tokens=False)
                if token_ids:
                    candidate_scores.append(float(logits[token_ids[0]].detach().cpu()))
            scores[letter] = max(candidate_scores) if candidate_scores else float('-inf')
        return max(scores, key=scores.get), scores
    except torch.cuda.OutOfMemoryError:
        torch.cuda.empty_cache()
        return None, {'error': f'CUDA OOM during {role} constrained choice'}
    except Exception as exc:
        return None, {'error': f'{type(exc).__name__}: {exc}'}

def direct_model_vote(question, role):
    qtext = format_question(question)
    prompt = f'''Answer this multiple-choice question.
Valid answers are only A, B, C, or D. Do not invent other options.
Return exactly: Final Answer: <A/B/C/D>

{qtext}

Final Answer:'''
    raw = generate_text(prompt, max_new_tokens=16, max_time=8, role=role)
    letter = extract_letter(raw)
    scores = None
    reason = 'parsed generated answer'
    if letter not in LETTERS:
        letter, scores = constrained_letter_choice_for_role(qtext, f'{role} raw output was invalid: {raw}', role=role)
        reason = 'constrained A-D logits for this model'
    return {
        'role': role,
        'letter': letter,
        'raw': raw,
        'scores': scores,
        'reason': reason,
    }

def majority_vote(votes):
    counts = {letter: 0 for letter in LETTERS}
    for vote in votes:
        if vote.get('letter') in LETTERS:
            counts[vote['letter']] += 1
    best_count = max(counts.values())
    winners = [letter for letter, count in counts.items() if count == best_count]
    if len(winners) == 1 and best_count > 0:
        return winners[0], counts, 'plain majority vote'

    score_totals = {letter: 0.0 for letter in LETTERS}
    for vote in votes:
        scores = vote.get('scores')
        if isinstance(scores, dict):
            for letter in LETTERS:
                value = scores.get(letter)
                if isinstance(value, (int, float)):
                    score_totals[letter] += value
    if any(score_totals.values()):
        return max(LETTERS, key=lambda letter: score_totals[letter]), counts, f'tie broken by score totals {score_totals}'

    for preferred in ['qwen', 'llama']:
        for vote in votes:
            if vote.get('role') == preferred and vote.get('letter') in LETTERS:
                return vote['letter'], counts, f'tie broken by {preferred} priority'
    return 'A', counts, 'emergency fallback'

def format_vote_debug(votes):
    parts = []
    for vote in votes:
        score_text = ''
        scores = vote.get('scores')
        if isinstance(scores, dict) and 'error' not in scores:
            score_text = ' scores=' + format_repair_scores(scores)
        elif isinstance(scores, dict) and 'error' in scores:
            score_text = f" error={scores['error']}"
        parts.append(f"{vote.get('role')}->{vote.get('letter')} ({vote.get('reason')}) raw={preview(vote.get('raw', ''), 200)}{score_text}")
    return '\n'.join(parts)

def print_debug_details(details):
    print('Votes:')
    print(details.get('vote_debug', ''))
    print('Vote counts:', details.get('vote_counts'))
    print('Selected letter:', details.get('selected_letter'), '| Reason:', details.get('selection_reason'))

def choose_answer(question):
    if len(question.options) < 4:
        return question.options[0].id, 'A', {'route': 'fallback', 'raw': 'Less than 4 options'}

    start_time = time.time()
    votes = [direct_model_vote(question, role) for role in ['qwen', 'llama']]
    letter, counts, reason = majority_vote(votes)
    idx = LETTERS.index(letter) if letter in LETTERS else 0
    details = {
        'route': 'two_model_qwen_llama',
        'raw': format_vote_debug(votes),
        'votes': votes,
        'vote_debug': format_vote_debug(votes),
        'vote_counts': counts,
        'code_output': 'Python code creation disabled; using preloaded Qwen + Llama 3B voting.',
        'selected_letter': LETTERS[idx],
        'selection_reason': reason,
        'elapsed_seconds': round(time.time() - start_time, 2),
    }
    return question.options[idx].id, LETTERS[idx], details
````

````python
preload_ensemble(warm_role='qwen', keep_loaded=True)

game = client.game.start(competition_id=COMPETITION_ID, mode='text')

while game.in_progress:
    question = game.current_question
    if question is None:
        break

    print('Level:', game.current_level)
    print(question.text)
    for i, opt in enumerate(question.options):
        print(f"{chr(65+i)}) {opt.text}")

    option_id, letter, details = choose_answer(question)
    print('Predicted:', letter, '| Route:', details.get('route'), '| Seconds:', details.get('elapsed_seconds'))
    if details.get('code_output'):
        print('Python output:', preview(details['code_output']))
    print('Raw output:', preview(details.get('raw', '')))
    print_debug_details(details)

    try:
        result = game.answer(option_id)
    except TimeoutError:
        print('Timed out')
        break
    except RateLimitError:
        print('Rate limited, waiting...')
        time.sleep(5)
        result = game.answer(option_id)

    print('Correct:', result.correct, '| Earned:', result.earned_amount)

    if result.game_over:
        break

    time.sleep(0.5)

print('Final earned:', game.earned_amount)
````
